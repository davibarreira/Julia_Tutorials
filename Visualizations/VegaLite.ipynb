```julia
import Pkg
Pkg.activate(".")
```

```julia
using VegaLite, VegaDatasets
using DataFrames
```

```julia
df = DataFrame(dataset("cars"))

df[!,:s] = map(x->rand([1,2]), df[!,:Origin])
df[!,:new] = lowercase.(df[!,:Origin]);
```

```julia
df |> @vlplot(
    :circle,
    x=:Horsepower,
    y=:Miles_per_Gallon,
    color=:new,
    width=300,
    height=300
)
```

```julia
df |> @vlplot(
    :circle,
    x=:Horsepower,
    y=:Miles_per_Gallon,
    color=:new,
    size=:Miles_per_Gallon,
    width=400,
    height=400
)
```

```julia
df |> @vlplot(
    :circle,
    x=:Horsepower,
    y=:Miles_per_Gallon,
    color=:new,
    size="s:o",
    width=400,
    height=400
)
```
