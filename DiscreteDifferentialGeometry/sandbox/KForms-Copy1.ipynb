## 1. Exterior Algebra
Let $V$ be a vector space. The inner product $\langle \cdot, \cdot \rangle : V \times V \to \mathbb R$ satisfies symmetry, linearity and positivity.
We can think of the inner product as performing a measurement, or a projection.

Exterior algebra introduces a new operator called the wedge product $\wedge$, which is analogous to the span operator.
The wedge is:
* antisymmetric ($a \wedge b = - b \wedge a$),
* $ a \wedge a = 0$
* associative $a \wedge ( b \wedge c) = (a \wedge b) \wedge c$.

```julia
@basis ℝ^3 E e
const origin = Point(0,0,0);
```

```julia
a = 2*e1 + 1*e2 + 0*e3
b = -1*e1 + 2*e2 + 0*e3

c = -1*e1 + 0*e2 + 1*e3

# Wedge properties
@show a ∧ b == - (b ∧ a)
@show a ∧ a == 0
@show a ∧ (b ∧ c) == (a∧b)∧c

# Consider the two eectors below. The plane formed by them is gieen by the wedge.
a = 2*e1 + 1*e2 + 0*e3
b = -1*e1 + 2*e2 + 0*e3

@show a ∧ b;
# The result is 5 e₁₂ which is equal to 5 e₁ ∧ e₂
# in other words, the wedge returns the area of 5 and oriented up.
```

```julia
function vecsegment(a)
    [Segment(origin,Point(a.v)),Ball(Point(a.v),0.1)]
end

function ngon(a,b)
    pa = Point(a.v)
    pb = Point(b.v)
    pab = Point((to(pa) + to(pb)).coords)
    Ngon(origin,pa,pab,pb)
end
v = viz(vecsegment(a),color=:blue)
viz!(vecsegment(b),color=:blue)
viz!(ngon(a,b),alpha=0.5,)
v
```
