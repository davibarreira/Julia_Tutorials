#  ACSets

An ACSet is an *attributed C-Set*.

```julia
using Pkg
Pkg.activate(".")
Pkg.add("MLStyle")
```

```julia
using GATlab, Catlab.CategoricalAlgebra

using MLStyle: @data
```

```julia
# Write down the schema for a weighted graph
@present MyGraphSch(FreeSchema) begin
  V::Ob
  E::Ob
  src::Hom(E,V)
  tgt::Hom(E,V)
  T::AttrType
  S::AttrType
  weight::Attr(E,T)
  shape::Attr(V,S)
end
```

```julia
# Construct the type used to store acsets on the previous schema
# We *index* src and tgt, which means that we store not only
# the forwards map, but also the backwards map.
@acset_type WeightedGraph(MyGraphSch, index=[:src,:tgt])
```

```julia
# Construct a weighted graph, with floats as edge weights
g = @acset WeightedGraph{Float64,Symbol} begin
  V = 4
  E = 5
  src = [1,1,1,2,3]
  tgt = [2,3,4,4,4]
  weight = [7.2, 9.3, 9.4, 0.1, 42.0]
  shape = [:star,:square,:circle,:circle]
end
```

```julia
add_part!(g, :E, src=4, tgt=1, weight=10);
```
