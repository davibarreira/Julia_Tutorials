```julia
using Pkg
Pkg.activate(".")
using ACSets
```

```julia
SchemaGraph = BasicSchema([:E,:V], [(:src,:E,:V),(:tgt,:E,:V)],
                          [:L,:W], [(:x,:V,:W),(:y,:V,:W),(:label,:V,:L),(:weight,:E,:W),(:radius,:V,:W)])

# Creates the ACData{:L,:W} struct
@acset_type AcData(SchemaGraph, index=[:src,:tgt])

acd = AcData{Symbol,Real}()
add_part!(acd, :V, label=:a, radius=1,x=1,y=2)
add_part!(acd, :V, label=:b, radius=1,x=2,y=2)
add_part!(acd, :V, label=:c, radius=1,x=1,y=1)
add_part!(acd, :V, label=:d, radius=1,x=2,y=1)

add_part!(acd, :E, src=1,tgt=2, weight=1)
add_part!(acd, :E, src=1,tgt=3, weight=1)
add_part!(acd, :E, src=4,tgt=2, weight=1)
```
