```julia
using CGPlots
using Distances
using Ripserer
using OptimalTransport
using StructArrays
using LinearAlgebra
```

```julia
function noisy_circle(n; r=1, noise=0.1)
    points = NTuple{2,Float64}[]
    for _ in 1:n
        θ = 2π * rand()
        push!(points, (r * sin(θ) + noise * rand(), r * cos(θ) + noise * rand()))
    end
    return points
end


c = StructArray(noisy_circle(100))

drawsvg(quickplot(c.:1,c.:2,size=c.:1,mark=Circle(r=2),figsize=(300,300)))
# drawsvg(quickplot(c.:1,c.:2,color=c.:1,size=c.:2,mark=Circle(r=1)),height=500)
```

```julia
M = pairwise(Euclidean(),c);
r = ripserer(M)

s = r[2][end].birth_simplex
```
