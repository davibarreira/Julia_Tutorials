```julia
ℤ_2ℤ, = residue_ring(ZZ, 2)

R = ℤ_2ℤ
⊕(x,y) = R(x) + R(y)
⊙(x,y) = R(x) * R(y)

R(0),R(1),R(2),R(3)
```
