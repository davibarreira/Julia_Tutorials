```julia
using Pkg
Pkg.activate(".")

using Distances
Pkg.add("Vizagrams")
using Vizagrams
using LinearAlgebra
```

```julia
p = map(x->[cos(x),sin(x)],range(0,2π,length=157))


rs = 0.1:0.1:1

metrics = [Euclidean(),SqEuclidean(),TotalVariation(),Cityblock(),Chebyshev(),Minkowski(1),Minkowski(1.5),Minkowski(0.5)]
d = mapreduce(metric->
    TextMark(text=string(metric),fontsize=0.1)+Line(map(x->x/evaluate(metric,[0,0],x),p)),
    (x,y)->y + T(-4,0)*x,
    metrics
)

drawsvg(d)
```

```julia
p = map(x->[cos(x),sin(x)],range(0,2π,length=100))
rs = 0.4:0.1:1
ps = map(r->map(x->r*[cos(x),sin(x)],range(0,2π,length=100)),rs)

metrics = [Euclidean(),SqEuclidean(),TotalVariation(),Cityblock(),Chebyshev(),Minkowski(1),Minkowski(1.5),Minkowski(0.5)]
d = mapreduce(zs->
        begin
        p = zs[1]
        r = zs[2]
        mapreduce(metric->
        Line(map(x->x/evaluate(metric,[0,0],x)*norm(x),p)),
        (x,y)->T(5,0)*x + y,
        reverse(metrics)
    ) + T(-3,0)TextMark(text=r,fontsize=0.4)
    end,
    (x,y)->y + T(0,-5)*x,
    zip(ps,rs))

drawsvg(d,height=1000)
```
