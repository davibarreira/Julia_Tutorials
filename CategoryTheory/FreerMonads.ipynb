```julia
struct C{T}
    x::T
end
myfunc(x::Int)::String = string(x)

# Define a type for Lan
struct Lan{T, A}
    ga::T # T{X}
    f# X->A
end
function Lan(ga::T,f::Function) where T
    t = T.name.wrapper
    A = isempty(typeof(ga).parameters) ? Any : typeof(ga).parameters[1]  # Extract the type A from ga
    return Lan{t, A}(ga, f)
end
# Note that if `f = identity` , then id:A->A thus ga::T{A}

# Lifting a parametric type to a Lan functor
function lan(ga::T) where T
    t = T.name.wrapper
    A = isempty(typeof(ga).parameters) ? Any : typeof(ga).parameters[1]  # Extract the type A from ga
    return Lan{t, A}(ga, identity)
end

fmap(f::Function, l::Lan) = Lan(l.ga, f ∘ l.f)

ga = [1, 2, 3]  # Sample input for ga
result = lan(ga)
fmap(myfunc,result)
```
