```julia
struct RZeroF end
struct ROneF end
struct RAddF{T}
    _1::T
    _2::T
end
struct RMultF{T}
    _1::T
    _2::T
end
RingF{T} = Union{RZeroF,ROneF,RAddF{T}, RMultF{T}}

fmap(f::Function, x::RZeroF) = RZeroF()
fmap(f::Function, x::ROneF)  = ROneF()
fmap(f::Function, x::RAddF{T}) where T = RAddF(f(x._1),f(x._2))
fmap(f::Function, x::RMultF{T}) where T = RMultF(f(x._1),f(x._2))
```

```julia
abstract type Ring end

struct RZero <: Ring end
struct ROne <: Ring end
struct RAdd <: Ring
    _1::Ring
    _2::Ring
end
struct RMult <: Ring
    _1::Ring
    _2::Ring
end

unfix(x::RZero) = RZeroF()
unfix(x::ROne)  = ROneF()
unfix(x::RAdd) = RAddF{Ring}(x._1,x._2)
unfix(x::RMult) = RMultF{Ring}(x._1,x._2)

alg1(a::RZeroF) = 0
alg1(a::ROneF)   = 1
alg1(a::RAddF{Int})   = a._1 + a._2
alg1(a::RMultF{Int})  = a._1 * a._2

alg2(a::RZeroF) = " hello "
alg2(a::ROneF)   = " bye "
alg2(a::RAddF{String})   = a._2 * a._1
alg2(a::RMultF{String})  = a._1 * a._2

cata(alg, a::Ring) = alg(fmap(x->cata(alg,x), unfix(a)))
```

```julia
expression = RAdd(RAdd(ROne(),ROne()),RMult(ROne(),RZero()))

cata(alg2,expression)
```

```julia
@data TreeF{a} begin
    LeafF()
    NodeF(Int,::a, ::a)
end
fmap(f::Function, x::LeafF) = x
fmap(f::Function, x::NodeF) = NodeF(x._1,f(x._2),f(x._3))
NodeF(x::Int,y::TreeF{T},z::TreeF{T}) where T = NodeF{TreeF{T}}(x,y,z) 

@data Tree begin
    Leaf()
    Node(Int,::Tree, ::Tree)
end

fix(x::LeafF)  = Leaf()
fix(x::NodeF)= Node(x._1,x._2,x._3)

NodeF(x::Int,y::Tree, z::Tree) = NodeF{Tree}(x::Int,y,z)
fix(NodeF(10,fix(LeafF{Int}()),fix(LeafF{Int}())))
```
