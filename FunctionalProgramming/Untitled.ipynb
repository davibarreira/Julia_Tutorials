```julia
@data FreeF{a} begin
    Pure(::a)
    FreeList(::FreeF{a})
end
fmap(f::Function, x::Pure) = Pure(f(x._1))
fmap(f::Function, x::FreeF) = free(fmap(y -> fmap(f, y), unfree(x)))
unfree(x::Pure) = x._1
unfree(x::FreeList) = List{FreeF}(x._1)
free(x::Vector) = FreeList(x._1)

fmap(f::Function, l::Vector) = map(f,l)
mcata(λ, ρ, x::Pure) = λ(x._1)
mcata(λ, ρ, x::FreeF) = ρ(fmap(y -> mcata(λ, ρ, y), unfree(x)))
cata(alg::Function, x::Pure) = x._1
cata(alg::Function, x::FreeF) = alg(fmap(y -> cata(alg, y), unfree(x)))

# The η and μ make FreeF a monad
η(x) = Pure(x)
μ(x::Pure{<:FreeF}) = x._1
μ(x::FreeF{<:FreeF}) = free(fmap(μ, unfree(x)))
```
